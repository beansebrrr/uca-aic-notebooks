```python
# SETUP — run me first (no bugs here)
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from skimage import data

Image.fromarray(data.chelsea()).save("photo.jpg")   # creates a sample file called photo.jpg

def load_image(path):
    return np.array(Image.open(path).convert("RGB"))

def show(*images):
    fig, axes = plt.subplots(1, len(images), figsize=(5 * len(images), 4))
    for ax, im in zip(np.atleast_1d(axes), images):
        ax.imshow(im); ax.axis("off")
    plt.show()

def swap_channels(img, order=(2, 0, 1)):
    return img[:, :, list(order)]

img = load_image("photo.jpg")                       # a working image for bugs 2–5
print("Setup done. The file on disk is called photo.jpg")
```

## Bug 5  ·  make the photo brighter (no error… but is it right?)

```python
# Convert RGB values to a 16-bit integer, clip them using np.clip(),
# then convert them back into an 8-bit integer.
brighter = np.clip(img.astype(np.uint16) + 100, 0, 255).astype(np.uint8)
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(img); ax[0].set_title("original")
ax[1].imshow(brighter); ax[1].set_title("brighter?")
plt.show()
```
