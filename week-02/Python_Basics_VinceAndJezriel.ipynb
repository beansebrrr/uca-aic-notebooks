# Conditionals

```python
mood_score = 0.1  # from a sentiment model: 0 is very sad, 1 is very happy
hour = 23         # time of day, 0 to 23

if mood_score > 0.7:
    playlist = "Upbeat dance"
elif mood_score >= 0.4:
    playlist = "Chill acoustic"
else:
    playlist = "Calm piano"

if hour >= 22 and playlist == "Upbeat dance":
    playlist = "Late night lo fi"

print(f"Mood {mood_score}, hour {hour}: playing {playlist}")
```

# For Loops

```python
loss = 2.0
history = []

for epoch in range(1, 6):           # epochs 1 to 5
    loss = loss * 0.7               # the model improves by 30%
    history.append(round(loss, 3))  # save the loss
    print(f"Epoch {epoch}: loss = {loss:.3f}")

print("Loss history:", history)

# Count how many epochs had loss below 1.0
good_epochs = 0
for value in history:
    if value < 1.0:
        good_epochs = good_epochs + 1
print("Epochs with loss below 1.0:", good_epochs)

# The same count with a list comprehension
print(len([v for v in history if v < 1.0]))

""" Pyramid of Stars """

height = 5
symbol = "[]" # Insert pyramid symbol here (just make sure it's a string)

for i in range(1, height+1):
    # Print the spaces needed to center the pyramid
    # Also taking into account the symbol's string length, again for symmetry
    # Disables the "next line" parameter
    print(" " * (((height) - i) * len(symbol)), end="")
    # Print the symbol
    print(symbol * (i + (i - 1)))
```

# Functions

```python
def accuracy(correct, total):
    """Return accuracy as a percentage."""
    return correct / total * 100

def normalise(values):
    """Scale numbers to the range 0 to 1."""
    low, high = min(values), max(values)
    return [(v - low) / (high - low) for v in values]

def label_from_score(score, threshold=0.5):
    """Turn a model's score into a label."""
    if score >= threshold:
        return "positive"
    return "negative"

print(accuracy(45, 50))                        # 90.0
print(normalise([10, 20, 30, 50]))             # [0.0, 0.25, 0.5, 1.0]
print(label_from_score(0.73))                  # positive
print(label_from_score(0.73, threshold=0.75))  # make this print negative


""" Function Swap """

def snake_case_to_camel_case(string: str) -> str:
  """Converts snake_case_strings into camelCaseStrings"""
  camel_case = ""
  # Split the words by the _ symbol and capitalize each word before merging
  for word in string.split("_"):
    camel_case += word.capitalize()
  return camel_case


print(snake_case_to_camel_case("Hee_hee_hoo_hoo"))
```

# Numpy

```python
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((8, 8))    # an 8 by 8 black image (all zeros)
img[2, 2] = 1             # left eye (row 2, column 2)
img[2, 5] = 1             # right eye (row 2, column 5)
img[5, 2:6] = 1           # a mouth: row 5, columns 2 to 5
img[4, 1] = img[4, 6] = 1 # corners of the smile

print("Shape:", img.shape)  # should print (8, 8)
print("Lit pixels:", int(img.sum()))
plt.imshow(img, cmap="magma")
plt.title("My first pixel art")
plt.show()

scores = np.array([0.62, 0.91, 0.47, 0.88, 0.73])

print("Mean score:", scores.mean())
print("Scores above 0.7:", scores[scores > 0.7])
print("As percentages:", scores * 100)

""" Own photo """

own_img = np.ones((8, 8))
own_img[2:4, 1:3] = 0   # Left eye
own_img[2:4, 5:7] = 0   # Right eye
own_img[4:6, 3:5] = 0   # Mouth center
own_img[5:7, 2] = 0     # Left mouth corner
own_img[5:7, 5] = 0     # Right mouth corner

print("Darkened pixels:", (own_img.size - own_img.sum()))
plt.imshow(own_img, cmap="magma")
plt.title("*Hisses*")
plt.show()
```

# Pandas

```python
import pandas as pd

songs = pd.DataFrame({
    "title":  ["Oud Nights", "Desert Wind", "Neon Souq", "Blue Dhow", "Pearl", "Sandstorm"],
    "genre":  ["khaleeji", "ambient", "pop", "jazz", "khaleeji", "pop"],
    "tempo":  [96, 70, 128, 110, None, 140],
    "streams_k": [540, 120, 980, 75, 610, 870],
})

print("Rows and columns:", songs.shape)
print(songs["streams_k"].mean())  # average streams

print("Missing values:")
print(songs.isna().sum())
songs["tempo"] = songs["tempo"].fillna(songs["tempo"].median()) # fill with the median

fast = songs[songs["tempo"] > 120]  # songs faster than 120 bpm
print(fast[["title", "tempo"]])

top = songs.sort_values("streams_k", ascending=False) # most streamed first
print("Top song:", top.iloc[0]["title"])

songs["popular"] = songs["streams_k"] > 500 # a new column: a label for AI!
print(songs.groupby("genre")["streams_k"].mean())

# Show minutes
songs["minutes"] = 3
print(songs[["title", "minutes"]])

# Plot a bar chart
songs["streams_k"].plot(kind="bar", title="Streams (thousands)")
```
